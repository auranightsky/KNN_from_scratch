```python
import numpy as np
import torch


# study hour
x = np.array([1, 2, 3, 4, 5])
# actual
y = np.array([0, 0, 0, 1, 1])

# make prediction
w = 1.0
b = -2

# Model calculates logits
z = w * x + b
print("model prediction : ",z)
```

#### convert login into probabilities

```python
def sigmoid(y_logit):
    return  1 / (1 + np.exp(-y_logit))
```

#### Binary cross-entropy BCE

```python
def binary_cross_entropy(y, y_pred_prob):
    return -(y * np.log(y_pred_prob) + (1 - y) * np.log(1 - y_pred_prob))
```

```python
y_pred_prob = sigmoid(z)
print("Prediction as Probabilities: ", y_pred_prob)


loss = binary_cross_entropy(y, y_pred_prob)
mean_loss = np.mean(loss)
print("BCE : ", mean_loss)
```

```python
y_pred = torch.tensor([0.9, 0.2, 0.8])
y_true = torch.tensor([1.0, 0.0, 1.0])

loss_fn = torch.nn.BCELoss()
loss = loss_fn(y_pred, y_true)
print("BCE :", loss)
```

#### Binary Cross-entropy with Logits BCEWithLogits

```python
def bce_with_logits(y, y_logit):
    # convert logit to probability
    y_pred_prob = sigmoid(y_logit)

    # calculate loss
    loss = binary_cross_entropy(y, y_pred_prob)
    return loss
```

```python
loss = bce_with_logits(y, z)
mean_loss = np.mean(loss)
print("BCEWithLogits : ", mean_loss)
```

#### Multiclass Cross-Entropy

```python
y = np.array([1, 0, 0])

# model prediction probability
y_pred = np.array([0.7, 0.2, 0.1])

loss = -np.sum(y * np.log(y_pred))

print("Multiclass Cross-Entropy:", loss)
```

```python

# Model's raw output
logits = torch.tensor([2.0, 1.0, 0.0])

# the actual
y = torch.tensor([1.0, 1.0, -1.0])

# way1
# create a function 
loss_fn = torch.nn.CrossEntropyLoss()

# way2
# remember that Pytorch will do everything for you 
# so do not pass the probability here just pass the logits(raw)
loss = torch.nn.functional.cross_entropy(
    logits.unsqueeze(0),
    y.unsqueeze(0)
)

loss2 = loss_fn(logits, y)

print("loss ", loss.item())
print("loss2 ", loss2.item())
```

#### Hinge Loss (SVM support vector machine)

```python
y = np.array([1, -1, 1])
y_pred = np.array([0.8, -0.5, 0.3])
hinge_loss = np.maximum(0, 1 - y * y_pred)
print("Hinge Loss: ", hinge_loss)
```

```python

y = np.array([1, -1, 1])
y_pred = np.array([0.8, -0.5, 0.3])

y = torch.tensor(y, dtype=torch.float32)
y_pred = torch.tensor(y_pred, dtype=torch.float32)

loss = torch.maximum(
    torch.tensor(0.0),
    1 - y * y_pred
)

print("Hinge Loss:", loss)
```
