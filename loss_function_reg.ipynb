# Regression Loss Functions: MSE, MAE, and Huber Loss

## What is a loss function?

A **loss function** is a mathematical function that tells us **how wrong a model's prediction is**.

For regression, the model predicts a numerical value such as:

- house price
- temperature
- age
- salary
- product price

The basic idea is:

```text
Actual value + Model prediction
              ↓
        Calculate error
              ↓
        Loss function
              ↓
        One loss value
```

A **smaller loss** generally means the predictions are closer to the actual values according to that particular loss function.

In this notebook we will study three common regression loss functions:

1. **MSE** — Mean Squared Error
2. **MAE** — Mean Absolute Error
3. **Huber Loss**

We will implement them using both **NumPy** and **PyTorch**.


# 1. Our Example Data

We will use the same actual values and predictions for all three loss functions so we can compare them.

```text
Actual values:     [10, 20, 30]
Predicted values:  [7,  18, 35]
```

Here:

- `y` = actual/true values
- `y_pred` = model predictions

The errors are:

$$
prediction - actual
$$

So:

$$
7-10=-3
$$

$$
18-20=-2
$$

$$
35-30=5
$$

Therefore:

```text
Errors = [-3, -2, 5]
```

Notice that an error can be **negative or positive**.

- Negative error → prediction is below the actual value.
- Positive error → prediction is above the actual value.


````python
import numpy as np

# Actual values (y)
y = np.array([10, 20, 30])

# Predicted values (y-hat)
y_pred = np.array([7, 18, 35])

# Calculate the prediction errors
error = y_pred - y

print("Actual:    ", y)
print("Prediction:", y_pred)
print("Error:     ", error)
````

# 2. MSE — Mean Squared Error

## What is MSE?

**MSE** stands for **Mean Squared Error**.

MSE measures prediction error by:

1. calculating the error
2. **squaring** each error
3. adding the squared errors
4. taking their **mean (average)**

### Formula

$$
MSE = \\frac{1}{n}\\sum_{i=1}^{n}(y_i-\\hat{y}_i)^2
$$

You may also see:

$$
MSE = \\frac{1}{n}\\sum_{i=1}^{n}(\\hat{y}_i-y_i)^2
$$

These give the same result because squaring removes the negative sign.

---

## Understanding every part of the formula

### $MSE$

The final **Mean Squared Error** value.

### $n$

The number of examples.

We have 3 examples:

$$
n=3
$$

### $i$

The index of the example.

For our three examples:

$$
i=1,2,3
$$

### $y_i$

The **actual value** for example $i$.

### $\\hat{y}_i$

The **predicted value** for example $i$.

The `hat` over $y$ means **prediction**.

### $y_i-\\hat{y}_i$

The prediction error:

$$
actual-prediction
$$

For our data:

$$
10-7=3
$$

$$
20-18=2
$$

$$
30-35=-5
$$

### $(y_i-\\hat{y}_i)^2$

The `²` means **square the error**.

So:

$$
3^2=9
$$

$$
2^2=4
$$

$$
(-5)^2=25
$$

The squared errors are:

```text
[9, 4, 25]
```

### $\\sum$

`Σ` means **summation** — add everything.

$$
9+4+25=38
$$

### $\\frac{1}{n}$

This means divide by the number of examples.

$$
\\frac{38}{3}=12.67
$$

Therefore:

$$
\\boxed{MSE\\approx12.67}
$$

### Why does MSE square the error?

There are two important reasons:

**1. Positive and negative errors should not cancel each other.**

Without squaring:

$$
-3+3=0
$$

That would incorrectly suggest there is no error.

After squaring:

$$
(-3)^2=9
$$

$$
3^2=9
$$

Both errors count as errors.

**2. Large errors are punished more strongly.**

For example:

$$
2^2=4
$$

but:

$$
10^2=100
$$

So MSE gives much more importance to large mistakes.

### When should we use MSE?

MSE is useful when:

- large errors should be strongly penalized
- outliers should have a strong influence on the loss
- we want a smooth loss function that is commonly used in regression

### Important disadvantage

Because errors are squared, MSE can be **very sensitive to outliers**.

For example, an error of 10 becomes 100 after squaring.


## MSE in NumPy

The formula is:

$$
MSE = \\frac{1}{n}\\sum_{i=1}^{n}(y_i-\\hat{y}_i)^2
$$

In NumPy we can write:

```python
mse = np.mean((y_pred - y) ** 2)
```

Mapping the formula to the code:

| Formula | Python |
|---|---|
| $y_i-\\hat y_i$ | `y_pred - y` |
| square $^2$ | `** 2` |
| mean | `np.mean(...)` |

`np.mean()` performs the averaging part of the formula.


````python
# MSE using NumPy

squared_error = (y_pred - y) ** 2
mse = np.mean(squared_error)

print("Squared errors:", squared_error)
print("MSE:", mse)
````

## MSE in PyTorch

The same mathematical calculation can be written with PyTorch:

```python
mse = torch.mean((y_pred - y) ** 2)
```

Because `torch.mean()` works with floating-point tensors, we create the values using `dtype=torch.float32`.


````python
import torch

# Actual values
y_torch = torch.tensor([10, 20, 30], dtype=torch.float32)

# Predicted values
y_pred_torch = torch.tensor([7, 18, 35], dtype=torch.float32)

# MSE
squared_error = (y_pred_torch - y_torch) ** 2
mse = torch.mean(squared_error)

print("Squared errors:", squared_error)
print("MSE:", mse.item())
````

# 3. MAE — Mean Absolute Error

## What is MAE?

**MAE** stands for **Mean Absolute Error**.

MAE measures the **average size of the prediction errors**.

Instead of squaring the errors, MAE takes the **absolute value**.

### Formula

$$
MAE = \\frac{1}{n}\\sum_{i=1}^{n}|y_i-\\hat{y}_i|
$$

---

## Understanding every part

### $n$

The number of examples.

Here:

$$
n=3
$$

### $y_i$

The actual value.

### $\\hat y_i$

The predicted value.

### $y_i-\\hat y_i$

The error.

Our errors are:

$$
[3,2,-5]
$$

### $|y_i-\\hat y_i|$

The vertical bars mean **absolute value**.

Absolute value means the **distance from zero**, ignoring whether the number is positive or negative.

So:

$$
|3|=3
$$

$$
|2|=2
$$

$$
|-5|=5
$$

The absolute errors are:

```text
[3, 2, 5]
```

### $\\sum$

Add the absolute errors:

$$
3+2+5=10
$$

### $\\frac{1}{n}$

Divide by the number of examples:

$$
\\frac{10}{3}=3.33
$$

Therefore:

$$
\\boxed{MAE\\approx3.33}
$$

### Why use absolute value?

Imagine two errors:

$$
-3
$$

and:

$$
+3
$$

They are both **3 units away from the correct value**.

If we simply added them:

$$
-3+3=0
$$

the errors would cancel.

Absolute value prevents this:

$$
|-3|+|3|=3+3=6
$$

### When should we use MAE?

MAE is useful when:

- we want an error measure that is easy to interpret
- we want the result in the original units of the target
- we do not want large errors to dominate the loss as strongly as they do in MSE
- we want less sensitivity to outliers than MSE

For example, if predicting age and:

$$
MAE=3.33
$$

we can interpret this as the predictions being about **3.33 years away from the actual values on average**.

### MSE vs MAE

For the same data:

$$
MSE\\approx12.67
$$

while:

$$
MAE\\approx3.33
$$

These numbers are different because the two loss functions measure error differently.


## MAE in NumPy

The formula:

$$
MAE = \\frac{1}{n}\\sum_{i=1}^{n}|y_i-\\hat{y}_i|
$$

becomes:

```python
mae = np.mean(np.abs(y_pred - y))
```

Mapping:

| Formula | NumPy |
|---|---|
| error | `y_pred - y` |
| absolute value | `np.abs(...)` |
| average | `np.mean(...)` |


````python
# MAE using NumPy

absolute_error = np.abs(y_pred - y)
mae = np.mean(absolute_error)

print("Absolute errors:", absolute_error)
print("MAE:", mae)
````

## MAE in PyTorch

The PyTorch version is:

```python
mae = torch.mean(torch.abs(y_pred - y))
```

Mapping:

| Mathematical operation | PyTorch |
|---|---|
| error | `y_pred - y` |
| absolute value | `torch.abs(...)` |
| average | `torch.mean(...)` |


````python
# MAE using PyTorch

absolute_error = torch.abs(y_pred_torch - y_torch)
mae = torch.mean(absolute_error)

print("Absolute errors:", absolute_error)
print("MAE:", mae.item())
````

# 4. Huber Loss

## What is Huber Loss?

**Huber Loss** combines ideas from **MSE** and **MAE**.

The key idea is:

```text
Small error  →  behave like MSE
Large error  →  behave more like MAE
```

This gives us a compromise between the two.

### Why do we need Huber Loss?

MSE strongly punishes large errors because it squares them.

MAE does not punish large errors as strongly because it grows linearly.

Huber Loss tries to get the benefits of both:

- for small errors, use a **quadratic/squared** behavior like MSE
- for large errors, use a **linear** behavior like MAE

This makes Huber Loss **less sensitive to outliers than MSE**, while still being smoother around zero than MAE.

---

## Huber Loss Formula

One common definition is:

$$
L_\\delta(a)=
\\begin{cases}
\\frac{1}{2}a^2 & \\text{if } |a|\\leq\\delta \\\\
\\delta\\left(|a|-\\frac{1}{2}\\delta\\right) & \\text{if } |a|>\\delta
\\end{cases}
$$

This looks complicated, so let's break it down.

### What is $a$?

`a` represents the **error**.

For example:

$$
a = prediction-actual
$$

or the opposite sign, depending on the convention being used.

The important thing is that the loss uses the size of the error.

### What is $\\delta$?

`δ` (delta) is a **threshold**.

It decides when we switch from the MSE-like part to the MAE-like part.

For example, if:

$$
\\delta=1
$$

then:

- error whose absolute value is 1 or less → use the first branch
- error whose absolute value is greater than 1 → use the second branch

### First branch: small errors

If:

$$
|a|\\leq\\delta
$$

we use:

$$
\\frac{1}{2}a^2
$$

This is the **squared-error part**, so it behaves similarly to MSE.

### Second branch: large errors

If:

$$
|a|>\\delta
$$

we use:

$$
\\delta\\left(|a|-\\frac{1}{2}\\delta\\right)
$$

This part grows **linearly** with the error, more like MAE.

---

## Small-error example

Let:

$$
a=0.5
$$

and:

$$
\\delta=1
$$

First check:

$$
|0.5|\\leq1
$$

This is true, so use:

$$
\\frac{1}{2}(0.5)^2
$$

First square:

$$
(0.5)^2=0.25
$$

Multiply by one-half:

$$
\\frac{1}{2}(0.25)=0.125
$$

So:

$$
L=0.125
$$

---

## Large-error example

Let:

$$
a=5
$$

and:

$$
\\delta=1
$$

Check:

$$
|5|>1
$$

This is true, so use:

$$
1\\left(5-\\frac{1}{2}(1)\\right)
$$

Calculate:

$$
\\frac{1}{2}(1)=0.5
$$

Then:

$$
5-0.5=4.5
$$

So:

$$
L=4.5
$$

Compare this with MSE-style squaring:

$$
5^2=25
$$

Huber gives `4.5` instead of `25`, so the large error has much less influence.

### When should we use Huber Loss?

Huber Loss is useful when:

- we want something between MSE and MAE
- we want to reduce the influence of large outliers
- we still want a smooth loss around small errors
- we are doing regression and the data may contain unusual/large errors

### Simple memory trick

```text
MSE:
error → square it

MAE:
error → absolute value

Huber:
small error → square
large error → linear
```


## Huber Loss in NumPy

NumPy does not require a special built-in loss function here; we can implement the formula directly.

We first calculate the error:

$$
a=y_{pred}-y
$$

Then check whether:

$$
|a|\\leq\\delta
$$

If true, use:

$$
\\frac12a^2
$$

Otherwise use:

$$
\\delta\\left(|a|-\\frac12\\delta\\right)
$$

For this example we will use:

$$
\\delta=1
$$


````python
# Huber Loss using NumPy

delta = 1.0

error = y_pred - y

# Small errors: 1/2 * error^2
# Large errors: delta * (|error| - 1/2 * delta)
huber_loss = np.where(
    np.abs(error) <= delta,
    (1/2) * error ** 2,
    delta * (np.abs(error) - (1/2) * delta)
)

# Mean Huber loss across all examples
huber = np.mean(huber_loss)

print("Errors:", error)
print("Huber loss for each example:", huber_loss)
print("Mean Huber Loss:", huber)
````

## Huber Loss in PyTorch

PyTorch provides a built-in function called:

```python
torch.nn.functional.huber_loss()
```

It calculates Huber Loss for us.

We will calculate the **mean** Huber loss over our examples.

The parameter `delta` controls the point where the behavior changes from squared to linear.

Here:

```python
delta = 1.0
```


````python
# Huber Loss using PyTorch

import torch.nn.functional as F

delta = 1.0

huber = F.huber_loss(
    y_pred_torch,
    y_torch,
    delta=delta,
    reduction="mean"
)

print("Mean Huber Loss:", huber.item())
````

# 5. Comparing MSE, MAE, and Huber

Now we can compare the three loss functions on the **same predictions**.

| Loss | Main idea | Large errors | Outlier sensitivity |
|---|---|---|---|
| **MSE** | Square the error | Punished strongly | High |
| **MAE** | Absolute value of error | Punished linearly | Lower |
| **Huber** | MSE-like for small errors, MAE-like for large errors | Moderate | Moderate |

### The easiest way to remember them

$$
\\boxed{MSE = error^2}
$$

$$
\\boxed{MAE = |error|}
$$

$$
\\boxed{Huber =
\\begin{cases}
\\text{square for small errors}\\\\
\\text{linear for large errors}
\\end{cases}}
$$

### Important point

There is no universal rule saying that one loss function is always the best.

The appropriate loss depends on what we want the model to care about.

For example:

- If very large errors should be heavily punished → **MSE** can be useful.
- If we want an easily interpretable average error and less outlier sensitivity → **MAE** can be useful.
- If we want a compromise between MSE and MAE → **Huber** can be useful.


````python
# Compare all three losses using PyTorch

mse = torch.mean((y_pred_torch - y_torch) ** 2)
mae = torch.mean(torch.abs(y_pred_torch - y_torch))
huber = F.huber_loss(y_pred_torch, y_torch, delta=1.0, reduction="mean")

print("MSE:  ", mse.item())
print("MAE:  ", mae.item())
print("Huber:", huber.item())
````

# 6. Quick Review

## MSE

**What?**

Mean Squared Error.

**Why?**

Squares errors, so large errors have a much stronger effect.

**When?**

Useful when large mistakes should be strongly penalized.

**Formula:**

$$
MSE = \\frac{1}{n}\\sum_{i=1}^{n}(y_i-\\hat y_i)^2
$$

---

## MAE

**What?**

Mean Absolute Error.

**Why?**

Uses absolute values, so positive and negative errors cannot cancel.

**When?**

Useful when we want an interpretable error in the original units and less sensitivity to outliers than MSE.

**Formula:**

$$
MAE = \\frac{1}{n}\\sum_{i=1}^{n}|y_i-\\hat y_i|
$$

---

## Huber

**What?**

A combination of squared-error behavior and absolute-error behavior.

**Why?**

It reduces the influence of large errors while keeping a smooth squared-error region for small errors.

**When?**

Useful when we want a compromise between MSE and MAE.

**Formula:**

$$
L_\\delta(a)=
\\begin{cases}
\\frac{1}{2}a^2 & |a|\\leq\\delta \\\\
\\delta\\left(|a|-\\frac{1}{2}\\delta\\right) & |a|>\\delta
\\end{cases}
$$

### Final memory trick

```text
MSE   → square the error
MAE   → absolute value of the error
Huber → square small errors, linear large errors
```
