```python
import torch
import matplotlib.pyplot as plt

# points
x_trains = torch.tensor([
    [2, 4],[1, 3],[2, 3],[3, 2],[2, 1],
    [5, 6],[4, 5],[4, 6],[6, 6],[5, 4]
],dtype=torch.float32)

# class/labels
y_train = torch.tensor([
    0, 0, 0, 0, 0,   # blue
    1, 1, 1, 1, 1    # red
])

# new point
x_test = torch.tensor([3.0, 3.0])
```

```python
class KNeareastNeighbors:
    def __init__(self, k=3):
        self.k = k

    def fit(self, x_train, y_train):
        self.x_train = x_train
        self.y_train = y_train

    def __euclidean_distance(self, x_test):
        return torch.sqrt(torch.sum((self.x_train - x_test)**2, dim=1))

    def predict(self, x_test):
        distances       = self.__euclidean_distance(x_test)
        sorted_indices  = torch.argsort(distances)
        nearest_indices = sorted_indices[:self.k]
        nearest_labels  = y_train[nearest_indices]

        return nearest_labels
```

```python
clf = KNeareastNeighbors()
clf.fit(x_trains, y_train)
print(clf.predict(x_test))
```
